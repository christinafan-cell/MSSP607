a. We will look at stock data from Facebook, Apple, Amazon, Netflix, and Google from different files.  We will combine them into a single file and store the dataframe for further analysis.

b. Read each file into a dataframe

```python
import pandas as pd
aapl = pd.read_csv("aapl.csv")
amzn = pd.read_csv("amzn.csv")
fb = pd.read_csv("fb.csv")
goog = pd.read_csv("goog.csv")
nflx = pd.read_csv("nflx.csv")
aapl.head()#to check if it works
```

c. Add a column to each dataframe, called 'ticker', indicating the ticker symbol it is for (Apple's is AAPL). This symbol is how you will look up a stock.  Each file name is the ticker symbol.

```python
aapl["ticker"] = "AAPL"
amzn["ticker"] = "AMZN"
fb["ticker"] = "FB"
goog["ticker"] = "GOOG"
nflx["ticker"] = "NFLX"
aapl.head()#to check if it works
```

d. Append each file together into a single dataframe.

```python
df = pd.concat([aapl, amzn, fb, goog, nflx])
df["ticker"].value_counts()#to check if it works
```

e. Save the results in a CSV file called 'faanf.csv'.

```python
df.to_csv("faanf.csv", index=False)
```

f. Using 'fanng.csv', use type conversion to change the 'date' column into a 'datetime' and the 'volume' column into integers.

```python
# Read the combined file
df = pd.read_csv("faanf.csv")

# Convert date to datetime
df["date"] = pd.to_datetime(df["date"])

# Convert volume to integer
df["volume"] = df["volume"].astype(int)

# Sort by date and ticker
df = df.sort_values(["date", "ticker"])

# Check data types
print(df.dtypes)

# Show first few rows
df.head()
```

g. Find the seven rows within the highest value for volume.

```python
top7 = df.nlargest(7, "volume")
print(top7)
```

h. Use the melt() method to make the 'date' and 'ticker' key identifiers.  Melt the remaining columns so that there are not separated.

```python
melted_df = df.melt(
    id_vars=["date", "ticker"],
    var_name="variable",
    value_name="value"
)
print(melted_df.shape)
print(melted_df.head(20))
```
