a. When does the code in a function execute: when the function is defined or called?

>The code in a function executes when the function is called, not when the function is defined. The def statement defines the function, while the code in the function's body runs when the function is called.



b. How many global scopes are there in a Python program? How many local scopes?
>There is only one global scope in a Python program. A new local scope is created whenever a function is called, so a program can have multiple local scopes. Multiple local scopes can exist at the same time when functions call other functions.

c. What happens to variables in a local scope when the function call returns?
>When a function returns, its local scope is destroyed, along with the local variables in that scope. Therefore, those local variables can no longer be accessed from outside the function.

d. How can you prevent a program from crashing when it gets an error?
>You can use try and except statements to handle errors. Code that might cause an error is placed in the try clause, and if an error occurs, Python moves to the except clause to handle the error instead of allowing the unhandled exception to crash the program.

Collatz Sequence Project

```python
def collatz(number):
  # If the number is even, divide it by 2.
  if number%2==0:
    print(number//2,end=' ')
    return number//2
  # If the number is odd, multiply it by 3 and add 1.
  else:
    print(number*3+1,end=' ')
    return number*3+1

# Ask the user to enter an integer.
number=int(input('Enter number:'))

# Print the starting number.
print(number, end=' ')

# Keep calling collatz() until the result is 1.
while number != 1:
    number = collatz(number)
```
