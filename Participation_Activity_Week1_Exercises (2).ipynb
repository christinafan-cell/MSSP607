## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
mean=sum(salaries)/len(salaries)
print(mean)
```

Verification:

```python
import statistics
print(statistics.mean(salaries))
```

### median

```python
sorted_salaries=sorted(salaries)
n=len(sorted_salaries)
median=(sorted_salaries[n//2-1]+sorted_salaries[n//2])/2
print(median)
```

Verification:

```python
import statistics
print(statistics.median(salaries))
```

### mode

```python
counts={}
for salary in salaries:
  counts[salary]=counts.get(salary,0)+1
mode=max(counts,key=counts.get)
print(mode)
```

Verification:

```python
import statistics
print(statistics.mode(salaries))
```

### sample variance
Remember to use Bessel's correction.

```python
mean=sum(salaries)/len(salaries)
sample_variance=sum((x-mean)**2 for x in salaries)/(len(salaries)-1)
print(sample_variance)
```

Verification:

```python
import statistics
print(statistics.variance(salaries))
```

### sample standard deviation
Remember to use Bessel's correction.

```python
sample_std=sample_variance**0.5
print(sample_std)
```

Verification:

```python
import statistics
print(statistics.stdev(salaries))
```

## Exercise 6: Calculating more statistics
### range

```python
data_range=max(salaries)-min(salaries)
print(data_range)
```

### coefficient of variation

```python
mean=sum(salaries)/len(salaries)
sample_std=(sum((x-mean)**2 for x in salaries)/(len(salaries)-1))**0.5
cv=sample_std/mean
print(cv)
```

### interquartile range

```python
import statistics
q1=statistics.quantiles(salaries,n=4)[0]
q3=statistics.quantiles(salaries,n=4)[2]
iqr=q3-q1
print("Q1:",q1)
print("Q3:",q3)
print("IQR:",iqr)
```

### quartile coefficent of dispersion

```python
quartile_coefficient=(q3-q1)/(q3+q1)
print(quartile_coefficient)
```

## Exercise 7: Scaling data
### min-max scaling

```python
min_salary=min(salaries)
max_salary=max(salaries)
scaled_salaries=[(x-min_salary)/(max_salary-min_salary)for x in salaries]
print(scaled_salaries)
```

### standardizing

```python
mean=sum(salaries)/len(salaries)
sample_std=(sum((x-mean)**2 for x in salaries)/(len(salaries)-1))**0.5
standardized_salaries=[(x-mean)/sample_std for x in salaries]
print(standardized_salaries)
```

Verification：

```python
mean=sum(salaries)/len(salaries)
sample_std=(sum((x-mean)**2 for x in salaries)/(len(salaries)-1))**0.5
standardized_salaries=[(x-mean)/sample_std for x in salaries]
print(standardized_salaries)
print("Mean:",sum(standardized_salaries)/len(standardized_salaries))
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
from statistics import mean, stdev
mean_salary,std_salary=mean(salaries),stdev(salaries)
standardized=[(x-mean_salary)/std_salary for x in salaries]
standardized[:5]
import numpy as np
covariance=np.cov(scaled_salaries,standardized)
print(covariance)
```

### Pearson correlation coefficient ($\rho$)

```python
import numpy as np
correlation=np.corrcoef(scaled_salaries,standardized)
print(correlation)
```
