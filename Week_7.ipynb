###1. Import file

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Upload the Internet Sales dataset
from google.colab import files

uploaded = files.upload()
```

```python
# Get the uploaded file name
file_name = list(uploaded.keys())[0]

# Read the CSV file
# latin1 is used because the dataset contains some non-UTF-8 characters
df = pd.read_csv(file_name, encoding='latin1', low_memory=False)
```

###2. Convert OrderDate to datetime format



```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
```

###3. Extract the year from OrderDate

```python
df['Year'] = df['OrderDate'].dt.year
```

###4. Calculate total sales for each year

```python
yearly_sales = df.groupby('Year')['SalesAmount'].sum()
# Display the yearly sales
print("Yearly Sales:")
print(yearly_sales)
```

###5. Create a bar chart

```python
plt.figure(figsize=(10, 6))

plt.bar(yearly_sales.index.astype(str), yearly_sales.values)

plt.title('Yearly Internet Sales')
plt.xlabel('Year')
plt.ylabel('Total Sales ($)')

plt.xticks(rotation=45)
plt.tight_layout()
```

###6. Save the active figure as a PDF file

```python
plt.savefig('yearly_sales.pdf', format='pdf', bbox_inches='tight')
```

###7. Display the bar chart

```python
plt.show()
```

###8. Download the PDF file

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(yearly_sales.index.astype(str), yearly_sales.values)

ax.set_title('Yearly Internet Sales')
ax.set_xlabel('Year')
ax.set_ylabel('Total Sales ($)')

plt.tight_layout()

# Display the chart
plt.show()

# Save the figure as PDF
fig.savefig('yearly_sales.pdf', format='pdf', bbox_inches='tight')

print("PDF file saved successfully!")
```
